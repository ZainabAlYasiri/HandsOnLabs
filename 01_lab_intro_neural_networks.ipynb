# Building a neural network

Neural networks are models inspired by the human brain. They are widely used in several areas, specially in computer vision tasks.

We will cover the basic concepts of neural networks and how to build basic ones for classification.


<img src="https://miro.medium.com/v2/resize:fit:1358/1*tWvofX1zZZnIHyHs-nKAfA.png" width="400">


A neural network is composed of layers of interconnected neurons (or nodes). Each neuron performs a weighted sum of its inputs, applies an activation function, and passes the result to the next layer. Neural networks are typically organized into three types of layers:

1. **Input Layer:** This layer receives the raw data, such as features or pixel values of an image.

2. **Hidden Layers:** These layers, which can be one or more, perform feature extraction and transformation. Each neuron in a hidden layer computes a weighted sum of its inputs and applies an activation function (e.g., ReLU, sigmoid) to introduce non-linearity.

3. **Output Layer:** The final layer produces the network's predictions. In classification tasks, the output layer often has one neuron per class and uses a softmax activation to generate class probabilities.

The first step for building the neural network is to define the architecture. There are multiple types of layers that you can use with different activation functions, units,etc; making the decision of the architecture highly diverse and a key part of the recipe for a good model.

How keras works is that you could create a "sequential" object and stack the layers within this. For example:
